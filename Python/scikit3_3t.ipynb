## Trade-off Bias & Variance
Bias merujuk pada kesalahan akibat asumsi yang terlalu sederhana, sedangkan varians mengukur sensitivitas terhadap fluktuasi kecil dalam data pelatihan. Menyeimbangkan keduanya adalah kunci untuk generalisasi.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r'E:\DigitalentDA\Datasets\advertising.csv')

print(df.head())
print(df.info())
print(df.describe())
```

```python
X = df[['TV', 'Radio', 'Newspaper']]
y = df['Sales']
```

```python
# Standarisasi Fitur
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Insialisasi model
linear_model = LinearRegression()
ridge_model = Ridge(alpha=10.0)
```

```python
# Evaluasi Kinerja dengan 5-Fold Cross Validation
linear_scores = cross_val_score(
    linear_model, X_scaled, y, cv=5,
    scoring='neg_mean_squared_error')
ridge_scores = cross_val_score(
    ridge_model, X_scaled, y, cv=5,
    scoring='neg_mean_squared_error')

print(f"RMSE Linear Regression: {-linear_scores.mean():.4f}")
print(f"RMSE Ridge Regression: {-ridge_scores.mean():.4f}")
```

```python
# Analisis Dampak Regularisasi pada Koefisien
linear_model.fit(X_scaled, y)
ridge_model.fit(X_scaled, y)

print("\n--- Perbandingan Koefisien (Bobot) Fitur ---")
fitur = ['TV', 'Radio', 'Newspaper']
for i, f in enumerate(fitur):
    print(f"Fitur {f:9}: Linear = {linear_model.coef_[i]:.4f} | Ridge = {ridge_model.coef_[i]:.4f}")
```
