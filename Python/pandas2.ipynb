## Summary Functions
Fungsi ringkasan adalah metode yang telah ditentukan sebelumnya dalam pustaka seperti Pandas yang membantu merangkum atau menggabungkan data. Fungsi ini memberikan wawasan cepat tentang karakteristik dataset, seperti rata-rata (mean), total (sum), rentang (range), dan distribusi. Fungsi-fungsi ini sangat penting dalam Exploratory Data Analysis (EDA) karena memungkinkan kita memahami data dengan sekilas.

### describe()
Fungsi describe() menghasilkan ringkasan statistik dari kolom numerik dalam dataset. Ini memberikan metrik seperti jumlah (count), rata-rata (mean), standar deviasi (std), minimum (min), maksimum (max), dan nilai kuartil.

```python
import pandas as pd
df1 = pd.read_csv('dataset1.csv')
df2 = pd.read_csv('dataset2.csv')
df3 = pd.read_csv('dataset3.csv')

print(df1.describe())
# print(df1.columns)
```

### mean()
Fungsi mean() menghitung nilai rata-rata dari sebuah kolom. Ini sangat berguna untuk memahami kecenderungan pusat dalam data numerik.

```python
avg_pts = df1['PTS'].mean()
print(f"Nilai rata-rata: {avg_pts}")
```

### sum()
Fungsi sum() menghitung total semua nilai dalam sebuah kolom. Ini berguna untuk analisis keuangan atau data kumulatif.

```python
total_salary = df1['SAL'].sum()
print(f"Total gaji pemain: {total_salary}")
```

### min() dan max()

```python
min_score = df1['SCORE'].min()
max_score = df1['SCORE'].max()
print(f"Score terkecil: {min_score}")
print(f"Score tertinggi: {max_score}")
```

### value_counts()
Menghitung frekuensi nilai unik

```python
counts_pos = df1['POS'].value_counts()
# print(f"Jumlah posisi: {counts_pos}")
print(counts_pos)
```

### unique()
Menampilkan nilai unik.

```python
unique_pos = df1['POS'].unique()
print(unique_pos)
```

## Maps
Dalam analisis data, "pemetaan" merujuk pada operasi yang menerapkan fungsi atau transformasi ke setiap elemen dalam dataset. Pemetaan memungkinkan kita memodifikasi atau meningkatkan data secara sistematis, membuatnya lebih cocok untuk analisis atau visualisasi.

### map()
Fungsi map() menerapkan transformasi ke setiap elemen dalam Series. Ini sering digunakan untuk transformasi sederhana, seperti mengonversi unit atau mengkodekan variabel kategorikal.

```python
print(df1.columns)
```

```python
df1['SCS'] = df1['PTS'].map(lambda x: (x / 5))
```

```python
print(df1.head())
```

```python
print(df2.columns)
```

```python
def categorize_attendances(attendances):
    if attendances > 40000 and attendances <= 60000:
        return 'Minor'
    elif attendances > 60000 and attendances <= 90000:
        return 'Normal'
    elif attendances > 90000:
        return 'Mayor'
    
df2['attendance_category'] = df2['attendance'].map(categorize_attendances)
```

```python
print(df2[['year', 'host_country', 'stadium', 'attendance', 'attendance_category']])
```

```python
print(df2['attendance_category'].value_counts())
```

```python
print(df2.groupby('attendance_category')['attendance'].mean())
```

```python
print(df2[['year', 'winner', 'winner_goals', 'runner_up', 'runner_up_goals']])
```

### apply()
Fungsi apply() lebih fleksibel daripada map(). Ini dapat diterapkan ke baris dan kolom DataFrame, serta mendukung operasi yang lebih kompleks.

```python
df1['SCS2'] = df1.apply(lambda row: row['PTS'] + row['REB'] + row['AST'] - row['STL'] - row['BLK'], axis=1)
```

## Grouping
Pengelompokan merujuk pada proses membagi dataset menjadi subset berdasarkan satu atau lebih variabel kategorikal. Setiap subset sesuai dengan kombinasi unik nilai dari kolom pengelompokan yang ditentukan. Setelah dikelompokkan, Anda dapat melakukan operasi agregasi (misalnya, jumlah, rata-rata, hitungan) pada setiap subset untuk mengekstrak statistik ringkasan. <br>

Sebagai contoh, jika Anda memiliki dataset transaksi penjualan, Anda mungkin mengelompokkan data berdasarkan "wilayah" dan "kategori produk" untuk menghitung total penjualan untuk setiap kombinasi variabel tersebut.

### Why Grouping is Important?
- Ringkasan Data Pengelompokan memungkinkan Anda menyusutkan dataset besar menjadi ringkasan yang mudah dikelola.
- Identifikasi Pola Dengan menggabungkan data, Anda dapat mengungkap tren dan hubungan dalam kelompok tertentu.
- Perbandingan antar Segmen Pengelompokan memungkinkan Anda membandingkan metrik di berbagai kategori atau segmen.

```python
grouped_df2 = df2.groupby('winner')['winner_goals'].agg(['sum'])

print(grouped_df2)
```

```python
grouped_df1 = df1.groupby('POS')['AST'].agg(['mean'])

print(grouped_df1)
```

```python
multig_df2 = df2.groupby(['continent_winner', 'winner'])['top_scorer_goals'].agg(['sum'])

print(multig_df2)
```

```python
reset_grouped = multig_df2.reset_index()

print(reset_grouped)
```

## Sorting
Pengurutan melibatkan pengaturan baris dataset dalam urutan tertentu berdasarkan satu atau lebih kolom. Operasi ini sangat berguna untuk memprioritaskan titik data, mengidentifikasi ekstrem (misalnya, nilai tertinggi atau terendah), atau sekadar membuat dataset lebih mudah dibaca. 

### Why Sorting is Important
- Organisasi Data: Pengurutan membantu mengorganisasi data secara logis, sehingga lebih mudah dianalisis.
- Menyoroti Ekstrem: Anda dapat dengan cepat mengidentifikasi entri teratas atau bawah dalam dataset.
- Persiapan Visualisasi: Data yang diurutkan sering kali menjadi dasar untuk membuat visualisasi yang efektif.

```python
sorted_df3 = df3.sort_values(by='Usia', ascending=True)

print(sorted_df3)

# ascending=True : kecil --> besar
# ascending=False : besar --> kecil
```

```python
multisort_df3 = df3.sort_values(by=['Nama', 'Usia'], ascending=[True, False])

print(multisort_df3)
```

### Grouping + Sorting

```python
grouped_df2 = df2.groupby('winner')['winner_goals'].agg(['sum'])
sorted_df2 = grouped_df2.sort_values(by='sum', ascending=False)
top_country = sorted_df2.head(3)

print(top_country)
```

- Pilih Kolom Relevan: Pilih kolom pengelompokan yang logis untuk tujuan analisis.
- Gunakan Beragam Agregasi: Tidak hanya menggunakan sum(), gunakan juga mean() atau count() bila perlu.
- Tangani Nilai Hilang: Pastikan dataset bersih sebelum pengelompokan atau pengurutan.
- Visualisasikan: Gunakan diagram batang, pie chart, atau heatmap untuk menyajikan hasil dengan lebih baik.
- Optimasi Performa: Seleksi hanya kolom yang dibutuhkan untuk mengurangi beban komputasi pada dataset besar.
