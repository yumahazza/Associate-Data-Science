```python
import numpy as np
import matplotlib.pyplot as plt
```

## Konsep Probabilitas

```python
# Probabilitas
# Kasus : Simulasi pelemparan koin

# Reproducibility
np.random.seed(0)

# Simulasi 10000 percobaan pelemparan koin
throws = np.random.randint(0, 2, size=10000)

# Menghitung jumlah gambar (1) dan angka (0)
count_gambar = np.sum(throws)
count_angka = len(throws) - count_gambar

# count_angka = np.sum(throws)
# count_gambar = len(throws) - count_angka

probilitas_gambar = count_gambar / len(throws)
probilitas_angka = count_angka / len(throws)

print(f"Probabilitas munculnya gambar: {probilitas_gambar:.4f}")
print(f"Probabilitas munculnya angka: {probilitas_angka:.4f}")
```

```python
# Independensi Kejadian
# Dua kejadian dikatakan independen jika kejadian satu 
# tidak memengaruhi probabilitas kejadian lainnya. 

# Reproducibility
np.random.seed(0)

# Simulasi 10000 percobaan pelemparan 2 dadu
throws = np.random.randint(1, 7, size=(10000, 2))

# Menghitung kejadian munculnya angka genap
count_even_genap = np.sum(throws[:, 0] % 2 == 0)
    
# Menghitung kejadian munculnya angka ganjil
count_even_ganjil = np.sum(throws[:, 1] % 2 != 0)

# Menghitung kejadian munculnya 2 angka yang jika dijumlahkan hasilnya genap
count_even_plus_genap = np.sum((throws[:-1, 0] + throws[1:, 0]) % 2 == 0)

probabilitas_genap = count_even_genap / len(throws)
probabilitas_ganjil = count_even_ganjil / len(throws)
probabilitas_plus_genap = count_even_plus_genap / (len(throws) - 1)

print(f"Probabilitas munculnya angka genap: {probabilitas_genap}")
print(f"Probabilitas munculnya angka ganjil: {probabilitas_ganjil}")
print(f"Probabilitas munculnya 2 angka yang jika dijumlahkan hasilnya genap: {probabilitas_plus_genap}")
```

```python
# Probabilitas bersyarat: P(A|B) = P(A ∩ B) / P(B)
# Probabilitas bersyarat adalah probabilitas kejadian B terjadi 
# jika kejadian A sudah terjadi.
# Kasus : Probabilitas muncul angka genap jika hasil lemparan dadu lebih dari 3

np.random.seed(0)

throws = np.random.randint(1, 7, size=10000)

count_even_genap = np.sum(throws % 2 == 0)

count_greater_than_3 = np.sum(throws > 3)

count_even_case = np.sum((throws % 2 == 0) & (throws > 3))

probabilitas_genap = count_even_genap / len(throws)

probabilitas_greater_than_3 = count_greater_than_3 / len(throws)

probabilitas_akhir = count_even_case / count_greater_than_3

print(f"Probabilitas munculnya angka genap: {probabilitas_genap}")
print(f"Probabilitas munculnya angka lebih dari 3: {probabilitas_greater_than_3}")
print(f"Probabilitas munculnya angka genap jika hasil lemparan dadu lebih dari 3: {probabilitas_akhir}")
```

```python
# Kovariansi dan Independensi
# Kovariansi mengukur tingkat ketergantungan linier antara dua variabel acak. 
# Jika dua variabel independen, kovariansinya akan nol. 
# Namun, kovariansi nol tidak selalu menyiratkan independensi.

np.random.seed(0)
X = np.random.normal(loc=0, scale=1, size=1000)  # Dataset X
Y = np.random.normal(loc=0, scale=1, size=1000)  # Dataset Y

covariance_matrix = np.cov(X, Y)  # Covariance matrix
covariance = covariance_matrix[0, 1]  # Extract covariance
print(f"Covariance: {covariance}")
# Output: ~-0.0301 (close to zero, indicating near-independence)

'''
- np.random.normal(loc=0, scale=1, size=1000) menghasilkan data dari distribusi normal dengan mean (loc) 0 dan deviasi standar (scale) 1.

- np.cov(X, Y) menghitung matriks kovariansi antara dua dataset X dan Y.

- Kovariansi diambil dari elemen [0, 1] matriks, yang menunjukkan hubungan linier antara X dan Y. 
Nilai mendekati nol berarti kedua variabel hampir independen.
'''
```

## Distribusi Probabilitas Umum

```python
# Distribusi Uniform
# Dalam distribusi uniform, semua hasil memiliki kemungkinan yang sama. 
# Misalnya, probabilitas setiap hasil dalam lemparan koin adil adalah 0.5.

np.random.seed(100)
uniform_data = np.random.uniform(0, 10, 10000)
plt.hist(uniform_data, bins=50, color='blue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()

'''
- np.random.uniform(0, 1, 1000) menghasilkan 1.000 angka acak yang tersebar merata antara 0 dan 1.
- plt.hist() membuat histogram untuk memvisualisasikan distribusi data. 
Histogram menunjukkan bahwa semua nilai memiliki frekuensi yang sama, sesuai dengan sifat distribusi uniform.
'''
```

```python
# Distribusi Binomial
# Distribusi binomial menggambarkan jumlah keberhasilan dalam sejumlah percobaan independen, 
# masing-masing dengan probabilitas keberhasilan yang sama.

binomial_data = np.random.binomial(n=10, p=0.5, size=10000)
plt.hist(binomial_data, bins=10, color='blue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()

'''
- np.random.binomial(n=10, p=0.5, size=1000) mensimulasikan 1.000 eksperimen binomial 
dengan 10 percobaan dan probabilitas keberhasilan 0.5.

- plt.hist() menampilkan distribusi hasil, yang cenderung berbentuk lonceng karena 
probabilitas keberhasilan sama (0.5).
'''
```

```python
# Distribusi Poisson
# Distribusi Poisson memodelkan jumlah kejadian yang terjadi 
# dalam interval waktu atau ruang yang tetap.

poisson_data = np.random.poisson(lam=10, size=1000)
plt.hist(poisson_data, bins=10, color='blue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```

```python
# Distribusi Normal
# Distribusi normal adalah kurva berbentuk lonceng yang dicirikan oleh mean dan deviasi standar.
normal_data = np.random.normal(loc=2, scale=2, size=1000)
plt.hist(normal_data, bins=30, color='blue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```

```python
# Distribusi Eksponensial
# Distribusi eksponensial memodelkan waktu 
# antara kejadian dalam suatu proses Poisson.

exponential_data = np.random.exponential(scale=1, size=1000)
plt.hist(exponential_data, bins=30, color='blue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```
