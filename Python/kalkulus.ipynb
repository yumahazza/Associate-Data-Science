## Fungsi Linear
Fungsi ini sering digunakan dalam masalah regresi.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(1, 8, 100)
y = 2 * x + 5

plt.plot(x, y, label='f(x) = 2x + 5')
plt.title("Linear Function")
plt.legend()
plt.show()
```

## Fungsi Kuadrat
Fungsi ini sering digunakan untuk memodelkan hubungan nonlinear sederhana.

```python
x = np.linspace(1, 8, 400)
y = x**2 - 8*x + 5

plt.plot(x, y, label='f(x) = x^2 - 8x + 5')
plt.title("Quadratic Function")
plt.legend()
plt.show()
```

## Fungsi Polinomial
Fungsi ini banyak digunakan dalam regresi polinomial.

```python
x = np.linspace(1, 8, 1000)
y = 0.1 * x**5 - 0.5 * x**2 + 2 * x - 1

plt.plot(x, y, label='f(x) = 0.1x^3 - 0.5x^2 + 2x - 1')
plt.title("Polynomial Function")
plt.legend()
plt.show()
```

## Fungsi Eksponensial
Sering digunakan untuk pemodelan pertumbuhan atau peluruhan.

```python
x = np.linspace(1, 8, 400)
y = 3**x
plt.plot(x, y, label='f(x) = 2^x')
plt.title("Exponential Function")
plt.legend()
plt.show()
```

## Fungsi Logaritma
Dalam aplikasi IoT, data sensor sering kali memiliki rentang nilai yang sangat besar. Fungsi logaritmik dapat digunakan untuk menormalisasi data:

```python
x = np.linspace(0.1, 10, 400)
y = np.log(x * 5)
plt.plot(x, y, label='f(x) = log(x * 5)')
plt.title("Logarithmic Function")
plt.legend()
plt.show()
```

## Fungsi Trigonometrik
Fungsi ini penting dalam pemrosesan sinyal dan analisis deret waktu.

```python
x = np.linspace(-10, 50, 1000)
y = np.cos(x)
plt.plot(x, y, label='f(x) = cos(x)')
plt.title("Trigonometric Function")
plt.legend()
plt.show()
```

## Turunan (Derivatives)
Turunan suatu fungsi mengukur laju perubahan fungsi tersebut. Sangat penting untuk optimasi, seperti pada algoritma gradient descent.

```python
def f(x):
    return 3 * x**3 - 6 * x**2 + x - 4

def first_derivative(x):
    return 9 * x**2 - 12 * x + 1

def second_derivative(x):
    return 18 * x - 12

x = np.linspace(-3, 5, 100)
plt.plot(x, f(x), label='f(x)')
plt.plot(x, first_derivative(x), label="f'(x)")
plt.plot(x, second_derivative(x), label="f''(x)")
plt.title("Derivatives of a Polynomial Function")
plt.legend()
plt.show()
```

## Gradient Descent
Gradient descent adalah algoritma optimasi untuk menemukan minimum suatu fungsi.

```python
def f(x):
    return x**2 - 4*x + 3

def df_dx(x):
    return 2*x - 4

def gradient_descent(init_x, learning_rate=0.1, iterations=10):
    x = init_x
    x_history = [x]
    for _ in range(iterations):
        grad = df_dx(x)
        x -= learning_rate * grad
        x_history.append(x)
    return x, x_history

x_min, x_history = gradient_descent(0.0)
x = np.linspace(-3, 5, 100)
y = f(x)

plt.plot(x, y, 'r-', label='f(x) = x^2 - 4x + 3')
plt.plot(x_history, f(np.array(x_history)), 'bo', label='Gradient Descent', markersize=3)
plt.title("Gradient Descent Visualization")
plt.legend()
plt.show()
```

## Integral
Integral merepresentasikan luas di bawah kurva suatu fungsi.

```python
def f(x):
    return x

def F(x):
    return x**2 / 2

x = np.linspace(0, 2, 400)
y = f(x)
F_y = F(x)

fig, ax = plt.subplots()
ax.plot(x, y, 'r', linewidth=2, label='f(x) = x')
ax.plot(x, F_y, 'g', linewidth=2, label='F(x) = x^2/2')

# Fill area between 0 and 1
ax.fill_between(x, y, where=[(xi >= 0) and (xi <= 1) for xi in x], color='skyblue', alpha=0.5)

ax.set_title("Integral Visualization")
ax.legend()
ax.grid(True)
plt.show()
```
